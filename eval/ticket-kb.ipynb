```python
import pandas as pd
```

```python
# %%
eval_df = pd.read_csv(
    "hf://datasets/bitext/Bitext-customer-support-llm-chatbot-training-dataset/"
    "Bitext_Sample_Customer_Support_Training_Dataset_27K_responses-v11.csv"
)

print(eval_df.shape)
print(eval_df["intent"].nunique())
```

```python
# %%
eval_set = (
    eval_df
    .groupby("intent")
    .sample(n=1, random_state=42)
    .reset_index(drop=True)
)

print("Evaluation queries:", len(eval_set))
```

```python
eval_set
```

```python
# %%
remaining = eval_df.drop(index=eval_set.index)

extra = remaining.sample(
    n=3,
    random_state=123
)

eval_set = pd.concat(
    [eval_set, extra],
    ignore_index=True
)
```

```python
eval_set
```

```python
# %%
eval_set.insert(
    0,
    "query_id",
    [f"Q{i:02d}" for i in range(1, len(eval_set) + 1)]
)
```
